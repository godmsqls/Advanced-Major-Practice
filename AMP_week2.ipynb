```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('GitHub_프사.jpeg')

b, g, r = cv2.split(image)

# b, g, r 값 확인하기
plt.imshow(b)
plt.axis('off')
plt.show()
```

```python
# 나눠진 채널을 다시 합치기 (정상 순서)
merged = cv2.merge([b, g, r])

# 순서를 바꾸면 색상이 달라짐
merged_wrong = cv2.merge([r, g, b])

# 아래 imshow에 변수 넣어서 확인!
plt.imshow(merged_wrong)
plt.axis('off')
plt.show()
```

```python
# 색을 쉽게 구분하기 위해 HSV 색상 공간 이용
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
h, s, v = cv2. split(hsv)

# h, s, v 값 넣어서 확인!
plt.imshow(h)
plt.axis('off')
plt.show()
```

```python
# 원하는 색상만 선택하려면?? 범위 선택
import numpy as np
lower_blue = np.array([100, 100, 50])
upper_blue = np.array([130, 255, 255])

mask = cv2.inRange(hsv, lower_blue, upper_blue)

# mask 값 넣어서 확인!
# plt.imshow(mask)
# plt.axis('off')
# plt.show()

# mask를 적용해서 원하는 색상만 남기기
result = cv2.bitwise_and(image, image, mask=mask)
plt.imshow(result)
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('color_balls.jpg')
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# 파란색만 추출하기
lower_blue = np.array([100, 100, 50])
upper_blue = np.array([130, 255, 255])

mask_blue = cv2.inRange(hsv, lower_blue, upper_blue)

# 초록색만 추출하기 -> 값 설정 다시 해야됨!!!!!!!!
lower_green = np.array([50, 50, 100])
upper_green = np.array([130, 255, 130])

mask_green = cv2.inRange(hsv, lower_green, upper_green)

#mask에 원하는 색 대입해서 추출!
result = cv2.bitwise_and(image, image, mask=mask_green)
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

plt.imshow(result_rgb)
plt.axis('off')
plt.show()
```

```python
# 처리한 이미지를 파일로 저장
cv2.imwrite('gray_image.png', gray_image)
cv2.imwrite('mask.png', mask)

success = cv2.imwrite('blue_result.png', result)

print(success)
```
